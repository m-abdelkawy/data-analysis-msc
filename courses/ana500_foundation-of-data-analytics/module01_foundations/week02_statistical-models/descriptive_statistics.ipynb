```python
import yfinance as yf
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
import seaborn as sns
```

```python
data = yf.download("^GSPC", start="2000-01-01", end="2026-01-01", auto_adjust=True)
annual_returns = data["Close"].resample("YE").last().pct_change().dropna()


print("Mean:", annual_returns.mean().item() * 100)
print("Median:", annual_returns.median().item() * 100)

display(annual_returns)
```

```python
# calculate mean and median
mean = annual_returns.mean().item()
median = annual_returns.median().item()

# create histogram
# plt.hist(annual_returns, bins=15)
plt.figure(figsize=(14, 9))
sns.histplot(data=annual_returns, bins=15, kde=True, edgecolor="black",linewidth=0.8, shrink=0.95)

# display mean and median on the histogram
plt.axvline(mean, linestyle="-", label=f"Mean: {mean:.2%}")
plt.axvline(median,linestyle="--", label=f"Median: {median:.2%}")

# format x-axis as percentage
plt.gca().xaxis.set_major_formatter(PercentFormatter(1))

# Labels and title
plt.xlabel("Annual Return")
plt.ylabel("Number of Years")
plt.title("S&P 500 Annual Returns")

# Add grid
#plt.grid()
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Show legend
plt.legend()

# plt.show()
```
